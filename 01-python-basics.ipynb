### 1.4 Exercise

Calculate the difference in area:

* Ask the user to input a value for `length`, 

* which serves as both the radius of a circle and the side length of a square. 

* Calculate the difference in area between the two shapes.

---

> Round the result to 2 decimal places.

```python
# 1. Define a function
def area_difference(input_length):
    """
    Calculate the difference in area
    """
    pi = 3.1415926535897932384
    area_circle = pi * input_length ** 2
    area_square = input_length ** 2
    difference = round(area_circle - area_square, 2)
    return difference

# 2. # Main Program
# User Input
length = float(input("Please enter the length: "))

# Calculate the difference in area
area_diff = area_difference(length)

# Print the difference
print("The difference in area is: ", area_diff)
```

## 2. Branching and Iteration

### 2.1 Conditional Branching

```python
# User Input
int_num = int(input("Please enter an int number: "))

# Conditional Branching
# "and" signifies the intersection of two conditions.
# "or" signifies the union of two conditions.
if -10 <= int_num <= 10 and int_num == 0:
    print("This is zero.")
elif int_num == -1 or int_num == 1:
    print("The absolute value of this number is one.")
else:
    print("I don't recognize this number.")
```

```python
# Initialize Counter
n = 0

# While Loop
# Execute if the condition is met.
# "break" is used to exit the current loop.
while n <= 5:
    n += 1
    print("I'm number", n)
    if n == 4:
        break
```

### 2.3 For Loop

```python
# For Loop

# range(5)
# Execute 5 times, starting from 0
# range(1, 7, 2)
# Starting from 1, count every 2 units, 
# up to (but not including) 7.
for n in range(4):
    print("I'm number", n+1)

# String Index
word = "example"
print(word[3])

# Alphabetical Indexing Format
# Use len() to calculate the length of a variable.
for index in range(len(word)):
    if word[index] == "m":
        print("This word contains an 'm'.")
        break

# Standard Format
for search in word:
    if search == "m":
        print("This word contains an 'm'.")
        break
```

### 2.4 Exercise

Escape the Lost Forest:

The game interface is a 20x7 grid, with the top and bottom two rows representing a forest, and the middle three rows representing a path.

Define the input range such that the user’s starting position is located on the path.

Allow the user to choose a direction to move. 

If the user turns right, the game should indicate that they are still in the forest. 

If the user turns right two or more times, any subsequent right turn will cause the player to have an emotional breakdown; a single left turn, however, will allow the player to successfully escape the forest.

```python
# 1. Define all functions
def range_check (input_num, min_value, max_value):
    """
    Check whether the input value "input_num" 
    falls within the range [min_value, max_value];
    if it is outside this range,
    prompt for input again 
    until the value falls within the range;
    return the final input value that lies within the range.
    """
    while True:
        if min_value <= input_num <= max_value:
            output_num = input_num
            break
        else:
            print(
                "Please ensure the input is between", 
                min_value, "and", max_value, "."
            )
            input_num = int(input("Please re-enter: "))
    return output_num

def forest_create (map_role, length, width):
    """
    Input the forest location and character coordinates, 
    then render the images of the forest and the character.
    The letter "o" represents a road.
    """
    for index in range(len(map_role)):
        if map_role[index] != "o":
            print(20 * "*")
        else:
            if width == index + 1:
                print((length - 1) * " " + "O")
            else:
                print(" ")

def path_check (input_path):
    """
    Check whether the input "input_path" is "left";
    if it is "right", prompt for re-entry, 
    and trigger a crash after two such attempts;
    if it is any other string, 
    display a prompt indicating that the input is invalid.
    """
    n = 0
    while True:
        if input_path == "left":
            output_str = "Congratulations on your departure!"
            break
        elif input_path == "right":
            n += 1
            print(
                "You failed to leave the forest; "
                "please enter a direction again!"
            )
            if n > 2:
                print("Aaaaaaaaaaaaa!!!!")
            input_path = input(
                "Please choose your direction: left or right"
            )
        else:
            input_path = input(
                "Please enter a valid path: left or right"
            )
    return output_str

# 2. # Main Program
# Initial User Coordinate Input
people_length = range_check(
    int(input("Please enter the x-coordinate: ")), 
    1, 
    20
)
people_width = range_check(
    int(input("Please enter the y-coordinate: ")), 
    3, 
    5
)

# Plotting of initial user coordinates
forest_create("xxoooxx", people_length, people_width)

# Escape the Lost Forest
people_path = path_check(
    input("Please choose your direction: left or right")
)
print(people_path)
```

## 3. 猜测、二分

求解立方根：

请用户输入10000以内的正整数，分别检索：

[1000, 9999]是否为某个整数的立方；

[100, 999]的立方根为多少，精度0.1；

[1, 99]的立方根为多少，精度0.001；

---
> 保留两位小数

```python
# 猜测检查法；
# 近似法；
# 二分搜索法[最优]；

"""1. 函数封装计算逻辑"""
def cube_check (input_num):
    """
    根据用户输入整数的范围进入不同分支；
    [1000, 9999]范围执行
    [100, 999]范围执行
    [1, 99]范围执行“二分法”；
    """
    while True:
        if 1000 <= input_num <= 9999:
            for cube in range(input_num):
                if cube ** 3 > input_num:
                    if (cube - 1) ** 3 == input_num:
                        output_num = cube - 1
                        break
                    else:
                        output_num = "输入项不为某个整数的立方"
                        break
            # 猜测检查法：
            # 从1开始猜测，每次+1，直到找到刚好超出边界的情况，检验前一个值是否满足；

            break
        elif 100 <= input_num <= 999:
            epsilon = 0.1
            step = 0.01
            cube = 1.0
            while abs(cube ** 3 - input_num) >= epsilon:
                cube += step
            output_num = round(cube, 2)
            # 近似法：
            # 从1开始遍历，每次增加一个step，直到刚好达到精度要求，输出此时的立方根；

            break
        elif 1 <= input_num <= 99:
            epsilon = 0.001
            low = 0
            high = input_num
            guess = (low + high) / 2
            while abs(guess ** 3 - input_num) >= epsilon:
                if guess ** 3 > input_num:
                    high = guess
                else:
                    low = guess
                guess = (low + high) / 2
            output_num = round(guess, 2)
            # 二分搜索法：
            # 最小值为0，最大值为输入值，猜测值为(low+high)/2；
            # 若猜测值带入，结果偏大，说明猜测值偏大，更新最大值为猜测值后计算新的猜测值；
            # 若猜测值带入，结果偏小，说明猜测值偏小，更新最小值为猜测值后计算新的猜测值；
            # 重复此过程，直到刚好达到精度要求，输出此时的立方根；

            break
        else:
            input_num = int(input("请输入[1, 9999]内的整数："))
    return output_num

"""2. 找出符合条件的立方根"""
print(cube_check(int(input("请输入[1, 9999]内的整数："))))
```

## 4. 元组、列表

更新学习数据库内容：

用户有一个原始数据库database_study存储学习情况；

当用户进行新的学习行为时对数据库内容进行更新并返回新的数据库；

```python
# 元组 | 不可更改
# 列表 | 可更改

"""1. 函数封装计算逻辑"""
# 输入内容检查
def input_check(time, item, duration):
    """
    检查用户输入的三个变量是否合法；
    若不合法，请用户重新输入；
    """
    # 检查输入时间是否合法
    while True:
        if time //10000 >= 2000:
            if 1 <= time %10000 //100 <= 12:
                if 1 <= time %100 <= 30:
                    break
                else:
                    time = int(input("请输入合法日期：例20260101"))
            else:
                time = int(input("请输入合法日期：例20260101"))
        else:
            time = int(input("请输入合法日期：例20260101"))

    # 检查输入项目是否合法
    while True:
        if item[:6] == "MITOCW":
            break
        else:
            item = input("请输入合法项目：例MITOCW…")

    # 检查输入时长是否合法
    while True:
        if 0 <= duration <= 24:
            break
        else:
            duration = int(input("请输入合法时长：例12"))

    return (time, item, duration)

# 数据库内容更新
def database_update (database_old, time, item, duration):
    """
    将传入的原始数据库[元组]，转存为列表数据；
    寻找输入的item是否存在，若存在，更新时间和时长，若不存在，添加新项目；
    按时间顺序排列后重新输出为元组数据返回；
    """
    # 检验输入项是否合理，不合理需重新输入；
    (time, item, duration) = input_check(time, item, duration)

    # 将输入的database_old元组转换为列表；
    database_list = [list(row) for row in database_old]

    # 检查新输入的学习内容是否已经存在；
    # 若存在，合并原本内容，若不存在，添加新内容；
    found = False
    for row in database_list:
        if row[1] == item:
            found = True
            row[0] = time
            row[2] += duration
            break
    if not found:
        database_list.append([time, item, duration])

    # 按照列表中每行第一个元素大小进行降序排序
    database_list.sort(key=lambda x: x[0], reverse=True)

    # 将操作后的database_list列表转换为元组；
    database_new = tuple(tuple(row) for row in database_list)
    return database_new

"""2. 学习内容数据库更新主体"""
# 原始学习内容数据库
database_study = (
    (20251013, "MITOCW6.0001", 23),
    (20261012, "MITOCW6.0002", 14)
)

# 学习新内容后更新数据库
database_study = database_update(
    database_study,
    int(input("请输入本次学习时间：")),
    input("请输入学习项目："),
    int(input("请输入本次学习时长："))
)

# 打印输出新学习内容数据库
print(database_study)

# 注意：
# T = ()                | 元组创建
# L = []                | 列表创建
# L.extend([5, 6])      | 拼接多个
# L = L + [7, 8]        | 连接列表
# L.remove(100)         | 按值删除首个匹配项
# del L[0]              | 按索引删除
# x = L.pop()           | 弹出末尾元素
# x = L.pop(0)          | 按索引弹出元素
# L.sort()              | 原地排序
# sorted(L)             | 返回新排序列
# L.reverse()           | 原地反转
# L[1:3]                | 切片 [start:end)
# L[:]                  | 克隆列表
```
